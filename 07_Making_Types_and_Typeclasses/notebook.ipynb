```haskell
data Shape = Circle Float Float Float | Rectangle Float Float Float Float
```

```haskell
:t Circle
:t Rectangle
```

```haskell
surface :: Shape -> Float
surface (Circle _ _ r) = pi * r ^ 2
surface (Rectangle x1 y1 x2 y2) = abs (x2 - x1) * abs (y2 - y1)
```

```haskell
surface (Circle 10 20 10)
surface (Rectangle 0 0 100 100)
```

```haskell
data Point = Point Float Float deriving (Show)
data Shape = Circle Point Float | Rectangle Point Point deriving (Show)
```

```haskell
Circle (Point 10 20) 10
Rectangle (Point 0 0) (Point 100 100)
```

```haskell
surface :: Shape -> Float
surface (Circle _ r) = pi * r ^ 2
surface (Rectangle (Point x1 y1) (Point x2 y2)) = abs (x2 - x1) * abs (y2 - y1)
```

```haskell
surface (Circle (Point 10 20) 10)
surface (Rectangle (Point 0 0) (Point 100 100))
```

```haskell
nudge :: Shape -> Float -> Float -> Shape
nudge (Circle (Point x y) r) a b = Circle (Point (x + a) (y + b)) r
nudge (Rectangle (Point x1 y1) (Point x2 y2)) a b = Rectangle (Point (x1 + a) (y1 + b)) (Point (x2 + a) (y2 + b))
```

```haskell
circle = Circle (Point 10 20) 10
circleMoved = nudge circle 5 10
circleMoved

surface circle
surface circleMoved
```

```haskell
rectangle = Rectangle (Point 0 0) (Point 100 100)
rectangleMoved = nudge rectangle 10 20
rectangleMoved

surface rectangle
surface rectangleMoved
```

```haskell
baseCircle :: Float -> Shape
baseCircle r = Circle (Point 0 0) r

baseRectangle :: Float -> Float -> Shape
baseRectangle width height = Rectangle (Point 0 0) (Point width height)
```

```haskell
nudge (baseCircle 10) 5 5
nudge (baseRectangle 100 200) 10 20
```

#### Record Syntax

```haskell
-- Person : firstname, lastName, age, height, phoneNumber, iceCreamFlavor
data Person = Person String String Int Float String String deriving (Show)
```

```haskell
let person = Person "John" "Doe" 30 180.5 "123-456-7890" "Vanilla"
person
```

```haskell
firstName :: Person -> String
firstName (Person firstName _ _ _ _ _) = firstName

lastName :: Person -> String
lastName (Person _ lastName _ _ _ _) = lastName

age :: Person -> Int
age (Person _ _ age _ _ _) = age

height :: Person -> Float
height (Person _ _ _ height _ _) = height

phoneNumber :: Person -> String
phoneNumber (Person _ _ _ _ phoneNumber _) = phoneNumber

flavor :: Person -> String
flavor (Person _ _ _ _ _ flavor) = flavor
```

```haskell
firstName person
lastName person
age person
height person
phoneNumber person
flavor person
```

```haskell
data Person = Person {firstName :: String
                    , lastName :: String
                    , age :: Int
                    , height :: Float
                    , phoneNumber :: String
                    , flavor :: String 
                    } deriving (Show)
```

```haskell
-- by this way, we can access the fields of the Person data type using the field names directly,
-- without needing to define separate functions for each field.

:t flavor
:t firstName
```

```haskell
person = Person "John" "Doe" 30 180.5 "123-456-7890" "Vanilla"
person
```

```haskell
Person {firstName = "John", lastName = "Doe", age = 30, height = 180.5, phoneNumber = "123-456-7890", flavor = "Vanilla"}
```

```haskell
data Vector a = Vector a a a deriving (Show)

vplus :: Num t => Vector t -> Vector t -> Vector t
(Vector x1 y1 z1) `vplus` (Vector x2 y2 z2) = Vector (x1 + x2) (y1 + y2) (z1 + z2)

vectMult :: Num t => Vector t -> t -> Vector t
(Vector x y z) `vectMult` m = Vector (x * m) (y * m) (z * m)

scalarMult :: Num t => Vector t -> Vector t -> t
(Vector x1 y1 z1) `scalarMult` (Vector x2 y2 z2) = x1 * x2 + y1 * y2 + z1 * z2
```

```haskell
Vector 1 2 3 `vplus` Vector 4 5 6
Vector 1 2 3 `vplus` Vector 4 5 6 `vplus` Vector 7 8 9
Vector 1 2 3 `vectMult` 2
Vector 1 2 3 `scalarMult` Vector 4 5 6
```

#### Derived Instances

```haskell
data Person = Person {firstName :: String
                    , lastName :: String
                    , age :: Int
                    } deriving (Eq, Show)
```

```haskell
john = Person "John" "Doe" 30
jane = Person "Jane" "Smith" 25
alice = Person "Alice" "Johnson" 35

alice == john
alice == alice
```

```haskell
let folks = [john, jane, alice]
john `elem` folks

omer = Person "Omer" "Yaslitas" 25
omer `elem` folks
```

```haskell
omer
show omer
```

```haskell
data Person = Person {firstName :: String
                    , lastName :: String
                    , age :: Int
                    } deriving (Eq, Show, Read)
```

```haskell
alice = Person "Alice" "Johnson" 35
"Alice is " ++ show alice
```

```haskell
read "Person {firstName = \"Alice\", lastName = \"Johnson\", age = 35}" :: Person
```

```haskell
read "Person {firstName = \"Alice\", lastName = \"Johnson\", age = 35}" == alice
```

#### Type Synonyms

```haskell
type String = [Char]
```

```haskell
phoneBook :: [(String, String)]
phoneBook = [("Alice", "123-456-7890"), ("Bob", "987-654-3210"), ("Charlie", "555-555-5555")]
```

```haskell
type PhoneBook = [(String, String)]
```

```haskell
type PhoneNumber = String
type Name = String
type PhoneBook = [(Name, PhoneNumber)]
```

```haskell
inPhoneBook :: Name -> PhoneNumber -> PhoneBook -> Bool
inPhoneBook name pnumber pbook = (name, pnumber) `elem` pbook
```

```haskell
inPhoneBook "Alice" "123-456-7890" phoneBook 
```

```haskell
inPhoneBook "Omer" "555-555-5555" phoneBook
```
