#### Pattern Matching

```haskell
lucky :: (Integral a) => a -> String
lucky 7 = "LUCKY NUMBER SEVEN!"
lucky x = "Sorry, you're out of luck, pal!"
```

```haskell
lucky 1
lucky 7
```

```haskell
sayMe :: (Integral a) => a -> String
sayMe 1 = "One!"
sayMe 2 = "Two!"
sayMe 3 = "Three!"
sayMe 4 = "Four!"
sayMe 5 = "Five!"
sayMe x = "Not between 1 and 5"
```

```haskell
sayMe 1
sayMe 5
sayMe 10
```

```haskell
-- Recursive factorial function
factorial :: (Integral a) => a -> a
factorial 0 = 1
factorial n = n * factorial(n - 1)
```

```haskell
factorial 0
factorial 6
```

```haskell
addVectors :: (Num a) => (a, a) -> (a, a) -> (a, a)
addVectors (x1, y1) (x2, y2) = (x1 + x2, y1 + y2)
```

```haskell
addVectors (1, 2) (3, 9)
```

```haskell
-- funcs for triples
first :: (a, b, c) -> a
first (x, _, _) = x

second :: (a, b, c) -> b
second (_, y, _) = y


third :: (a, b, c) -> c
third (_, _, z) = z
```

```haskell
first (1, 2, 3)
second (1, 2, 3)
third (1, 2, 3)
```

```haskell
-- Pattern matching in list comprehension
let xs = [(1, 3), (4, 3), (2, 7)]
[a + b | (a, b) <- xs]
```

```haskell
head' :: [a] -> a
head' [] = error "Can't call head on an empty list"
head' (x:_) = x

firstTwo :: [a] -> [a]
firstTwo (x:y:_) = [x, y]
firstTwo _ = error "Can't call firstTwo on a list of size less than 2"
```

```haskell
head' [1, 2, 3]
head' []
```

```haskell
firstTwo [1, 2, 3]
firstTwo [1]
```

```haskell
tell :: (Show a) => [a] -> String
tell [] = "The list is empty"
tell [x] = "The list has one element: " ++ show x
tell [x,y] = "The list has two elements: " ++ show x ++ " and " ++ show y
tell (x:y:_) = "The list is long. The first two elements are: " ++ show x ++ " and " ++ show y
```

```haskell
tell ""
tell "H"
tell "Hi"
tell "Hello"
```

```haskell
-- Length by recursion
length' :: (Num b) => [a] -> b
length' [] = 0
length' (_:xs) = 1 + length' xs
```

```haskell
length' "Hello"
```

```haskell
-- Sum by recursion
sum' :: (Num a) => [a] -> a
sum' [] = 0
sum' (x:xs) = x + sum' xs
```

```haskell
sum' [1, 2, 3, 4]
```

```haskell
-- Patterns
capital :: String -> String
capital "" = "Empty string"
capital all@(x:xs) = "The first letter of " ++ all ++ " is: " ++ show x
```

```haskell
capital "Hello"
```

#### Guards

```haskell
densityTell :: (RealFloat a) => a -> String
densityTell density
    | density < 1.2 = "FLY!"
    | density < 1000.0 = "Have fun swimming"
    | otherwise = "SINK!"
```

```haskell
densityTell 500
densityTell 0.5
densityTell 3000
```

```haskell
densityTell :: (RealFloat a) => a -> a -> String
densityTell mass volume
    | mass / volume < 1.2 = "FLY!"
    | mass / volume < 1000.0 = "Have fun swimming"
    | otherwise = "SINK!"
```

```haskell
densityTell 1000 2
densityTell 20 40
densityTell 6000 2
```

```haskell
-- Max with guards
max' :: (Ord a) => a -> a -> a
max' a b
    | a > b = a
    | otherwise = b
```

```haskell
max' 1 2
max' 10 5
```

```haskell
-- Compare with guards
myCompare :: (Ord a) => a -> a -> Ordering
a `myCompare` b
    | a > b = GT
    | a == b = EQ
    | otherwise = LT
```

```haskell
myCompare 2 3
3 `myCompare` 1
```

#### Where

```haskell
densityTell :: (RealFloat a) => a -> a -> String
densityTell mass volume
    | density < air = "FLY!"
    | density < water = "Have fun swimming"
    | otherwise = "SINK!"
    where 
        density = mass / volume
        air = 1.2
        water = 1000.0
```

```haskell
densityTell 1000 2
densityTell 20 40
densityTell 6000 2
```

```haskell
initials :: String -> String -> String
initials firstName lastName = [f] ++ ", " ++ [l] ++ "."
    where
        (f:_) = firstName
        (l:_) = lastName
```

```haskell
initials "Omer" "Yaslitas"
```

```haskell
-- Define funcs in where blocks
calcDensity :: (RealFloat a) => [(a, a)] -> [a]
calcDensity xs = [density m v | (m, v) <- xs]
    where density mass volume = mass / volume
```

```haskell
calcDensity [(3, 2), (10, 2)]
```

```haskell
calcDensity' :: (RealFloat a) => (a,a) -> a
calcDensity' (m, v) = density m v
    where density mass volume = mass / volume
```

```haskell
calcDensity' (3, 2)
```

#### Let

```haskell
cylinderSurface :: (RealFloat a) => a -> a-> a
cylinderSurface r h =
    let sideArea = 2 * pi * r * h
        topArea = pi * r ^2
    in  sideArea + 2 * topArea
```

```haskell
cylinderSurface 2 5
```

#### Case Expressions

```haskell
head' :: [a] -> a
head' [] = error "Empty list"
head' (x:_) = x
```

```haskell
head' :: [a] -> a
head' xs = case xs of
                [] -> error "Empty list"
                (x:_) -> x
```

```haskell
head' "Hello"
```

```haskell
describeList :: [a] -> String
describeList xs = "The list is " ++ case xs of  [] -> "an empty list."
                                                [x]-> "a singleton list."
                                                xs -> "a long list."
```

```haskell
describeList []
describeList [2]
describeList "Hello"
```
