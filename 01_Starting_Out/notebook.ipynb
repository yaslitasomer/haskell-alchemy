#### Lists

```haskell
numbers = [1, 2, 3, 4, 5]
numbers
```

```haskell
-- Putting lists together
[1, 2, 3] ++ [4, 5, 6]
"Hello" ++ " " ++ "World"
```

```haskell
"A" ++ " small cat"
5 : [6, 7, 9]
```

```haskell
-- Get elements
"Hello" !! 1
[9.4, 33.2, 96.2] !! 1
```

```haskell
[[1, 2, 3], [4, 5, 6]] !! 1
```

```haskell
-- Comparing Lists
[3, 2, 1] > [2, 1, 0]
[3, 2, 1] > [2, 10, 100]
[3, 4, 2] < [3, 4]
```

```haskell
-- Some functions on lists
head [5, 4, 3, 2, 1]
tail [5, 4, 3, 2, 1]
last [5, 4, 3, 2, 1]
init [5, 4, 3, 2, 1]
length [5, 4, 3, 2, 1]
null [1, 2, 3]
null []
reverse [5, 4, 3, 2, 1]
```

```haskell
take 3 [5, 4, 3, 2, 1]
take 1 [3, 9, 2, 1, 5, 6]
take 5 [1, 2]
take 0 [6, 6, 6, 6, 6]
```

```haskell
drop 3 [8, 4, 2, 1, 5, 6]
drop 0 [1, 2, 3, 4]
drop 100 [1, 2, 3, 4]
```

```haskell
maximum [1, 9, 2, 3, 4]
minimum [1, 9, 2, 3, 4]
```

```haskell
sum [1, 2, 3, 4]
product [1, 2, 3, 4]
```

```haskell
elem 1 [1, 2, 3, 4]
1 `elem` [1, 2, 3, 4]

elem 2 [1, 3, 4]
```

#### Ranges

```haskell
[1..20]
['a'..'z']
['a'..'k']
```

```haskell
[2, 4..20]
[3, 6..20]
[20, 19..1]
```

```haskell
[0.1, 0.3..1] -- Avoid floating point numbers in ranges, as they can be inaccurate 
```

```haskell
take 24 [13, 26..] -- Infinite list, take first 24 elements
```

```haskell
take 10 (cycle [1, 2, 3])
take 20 (cycle "LOL ")
```

```haskell
take 12 (repeat 3)
replicate 12 3
```

#### List Comprehensions

```haskell
[x*2 | x <- [1..10]]
```

```haskell
[x*2 | x <- [1..10], x*2 >= 12]
```

```haskell
[x | x <- [1..10], x `mod` 3 == 1]
```
