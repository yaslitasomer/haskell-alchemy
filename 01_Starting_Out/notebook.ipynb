#### List Comprehensions

```haskell
[x*2 | x <- [1..10]]
```

```haskell
[x*2 | x <- [1..10], x*2 >= 12]
```

```haskell
[x | x <- [1..10], x `mod` 3 == 1]
```

```haskell
boomBangs xs = [if x < 10 then "BOOM!" else "BANG!" | x <- xs, odd x]
```

```haskell
boomBangs [7..13]
```

```haskell
[x | x <- [1..30], x `mod` 3 == 1, x `mod` 2 == 1]
```

```haskell
[x * y | x <- [2, 3, 4], y <- [5, 6, 7]]
```

```haskell
[x * y | x <- [2, 3, 4], y <- [5, 6, 7], x * y > 15]
```

```haskell
nouns = ["car", "house", "tree"]
adjectives = ["big", "small", "tall"]
[adjective ++ " " ++ noun | adjective <- adjectives, noun <- nouns]
```

```haskell
[1 | x <- [1..10], even x]
```

```haskell
sum [ 1 | x <- [1..10], even x]
```

```haskell
length' xs = sum [1 | _ <- xs]
```

```haskell
length' [1, 2, 3, 4, 5]
```

```haskell
removeUppercase st = [c | c <- st, c `elem` ['a'..'z'] || c == ' ']
```

```haskell
removeUppercase "Hello World!, this is a test string with Uppercase and lowercase letters."
```

```haskell
-- Nested list comprehensions
xxs = [[1, 3, 5, 2, 3, 1, 2, 4, 5], [1, 2, 3, 4, 5], [1, 2, 4, 2, 1]]
[[x | x <- xs, even x] | xs <- xxs]
```

#### An example about triangles 

```haskell
triangles = [(a, b, c) | c <- [1..10], b <- [1..c], a <- [1..b]]
triangles
```

```haskell
rightTriangles = [(a, b, c) | c <- [1..10], b <- [1..c], a <- [1..b], a^2 + b^2 == c^2]
rightTriangles
```

```haskell
rightTriangles' = [(a, b, c) | c <- [1..10], b <- [1..c], a <- [1..b], a^2 + b^2 == c^2, a + b + c == 24]
rightTriangles'
```
