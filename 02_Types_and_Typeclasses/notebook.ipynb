#### Type Declaration in Functions

```haskell
removeUppercase :: [Char] -> [Char]
removeUppercase st = [c | c <- st, c `elem` ['a'..'z'] || c `elem` ['0'..'9'] || c `elem` [' ', '.', ',', '!', '?', ';', ':', '-', '_', '(', ')', '[', ']', '{', '}', '"', '\'']]

removeUppercase "Hello, Haskell! This is a Test String."
```

```haskell
addFour :: Int -> Int -> Int -> Int -> Int
addFour a b c d = a + b + c + d
addFour 1 2 3 4
```

```haskell
:t addFour
```

#### Common Types

```haskell
factorial :: Integer -> Integer
factorial n = product [1..n]
```

```haskell
factorial 50
:t factorial 50
```

```haskell
-- Float is a single-precision floating-point number
circumference :: Float -> Float
circumference r = 2 * pi * r
circumference 5.0
:t circumference 5.0
```

```haskell
-- Double is a double-precision floating-point number
circumference' :: Double -> Double
circumference' r = 2 * pi * r
circumference' 5.0
:t circumference' 5.0
```
